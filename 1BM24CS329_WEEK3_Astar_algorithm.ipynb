```python
import heapq

# Goal state
goal_state = [[1,2,3],
              [4,5,6],
              [7,8,0]]  # 0 = blank

# Flatten for easier comparison
goal_flat = sum(goal_state, [])

def misplaced_tiles(state):
    """Heuristic: count misplaced tiles"""
    flat = sum(state, [])
    return sum(1 for i in range(9) if flat[i] != 0 and flat[i] != goal_flat[i])

def get_neighbors(state):
    """Generate neighbors by sliding blank"""
    neighbors = []
    n = 3
    flat = sum(state, [])
    idx = flat.index(0)
    x, y = divmod(idx, n)

    moves = [(x-1,y),(x+1,y),(x,y-1),(x,y+1)]
    for nx, ny in moves:
        if 0 <= nx < n and 0 <= ny < n:
            new_flat = flat[:]
            new_idx = nx*n + ny
            new_flat[idx], new_flat[new_idx] = new_flat[new_idx], new_flat[idx]
            neighbors.append([new_flat[i:i+n] for i in range(0,9,n)])
    return neighbors

def astar_misplaced(start):
    """A* search using misplaced tiles heuristic"""
    frontier = []
    heapq.heappush(frontier, (misplaced_tiles(start), 0, start, []))
    visited = set()

    while frontier:
        f, g, state, path = heapq.heappop(frontier)
        flat = tuple(sum(state, []))

        if flat in visited:
            continue
        visited.add(flat)

        if state == goal_state:
            return path + [state]

        for neighbor in get_neighbors(state):
            new_g = g + 1
            h = misplaced_tiles(neighbor)
            heapq.heappush(frontier, (new_g + h, new_g, neighbor, path + [state]))
    return None

# Example run
start_state = [[1,2,3],
               [4,0,6],
               [7,5,8]]

solution = astar_misplaced(start_state)
print("Solution path (misplaced tiles):")
for step in solution:
    for row in step:
        print(row)
    print()
```

```python
def manhattan_distance(state):
    """Heuristic: sum of Manhattan distances of tiles"""
    flat = sum(state, [])
    dist = 0
    for i, val in enumerate(flat):
        if val != 0:
            goal_idx = goal_flat.index(val)
            x1, y1 = divmod(i, 3)
            x2, y2 = divmod(goal_idx, 3)
            dist += abs(x1 - x2) + abs(y1 - y2)
    return dist

def astar_manhattan(start):
    """A* search using Manhattan distance heuristic"""
    frontier = []
    heapq.heappush(frontier, (manhattan_distance(start), 0, start, []))
    visited = set()

    while frontier:
        f, g, state, path = heapq.heappop(frontier)
        flat = tuple(sum(state, []))

        if flat in visited:
            continue
        visited.add(flat)

        if state == goal_state:
            return path + [state]

        for neighbor in get_neighbors(state):
            new_g = g + 1
            h = manhattan_distance(neighbor)
            heapq.heappush(frontier, (new_g + h, new_g, neighbor, path + [state]))
    return None

# Example run
solution2 = astar_manhattan(start_state)
print("Solution path (Manhattan distance):")
for step in solution2:
    for row in step:
        print(row)
    print()
```
