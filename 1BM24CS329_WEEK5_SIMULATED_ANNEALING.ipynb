```python

import random
import math

def simulated_annealing(initial_state, cost, neighbour,
                        T=1000, cooling_rate=0.95):

    current = initial_state

    while T > 0.1:

        # Generate a random neighbour
        next_state = neighbour(current)

        # Calculate change in cost
        delta_E = cost(current) - cost(next_state)

        # Accept better solution
        if delta_E > 0:
            current = next_state

        else:
            # Accept worse solution with probability
            p = math.exp(delta_E / T)

            if random.random() < p:
                current = next_state

        # Decrease temperature
        T = T * cooling_rate

    return current


# Example usage
def cost(x):
    return x ** 2

def neighbour(x):
    return x + random.uniform(-5, 5)

initial_state = 10

result = simulated_annealing(
    initial_state,
    cost,
    neighbour
)

print("Best solution:", result)
print("Minimum cost:", cost(result))
```
