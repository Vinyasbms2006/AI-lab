```python

# Truth Table Enumeration Algorithm

def pl_true(sentence, model):
    return sentence(model)


# Recursive function
def tt_check_all(kb, alpha, symbols, model):

    if len(symbols) == 0:

        if pl_true(kb, model):
            return pl_true(alpha, model)
        else:
            return True

    else:
        P = symbols[0]
        rest = symbols[1:]

        return (
            tt_check_all(kb, alpha, rest,
                         {**model, P: True})
            and
            tt_check_all(kb, alpha, rest,
                         {**model, P: False})
        )


# Main entailment function
def tt_entails(kb, alpha, symbols):
    return tt_check_all(kb, alpha, symbols, {})


# Knowledge Base
def KB(model):

    P = model['P']
    Q = model['Q']
    R = model['R']

    s1 = (not Q) or P
    s2 = (not P) or (not Q)
    s3 = Q or R

    return s1 and s2 and s3


# Query
def alpha(model):
    return model['R']


# Generate truth table
def generate_truth_table():

    symbols = ['Q', 'P', 'R']

    print("\nTRUTH TABLE")
    print("-" * 75)

    print("Q\tP\tR\tQ->P\tP->~Q\tQvR\tKB")

    print("-" * 75)

    for q in [True, False]:
        for p in [True, False]:
            for r in [True, False]:

                model = {'Q': q, 'P': p, 'R': r}

                s1 = (not q) or p
                s2 = (not p) or (not q)
                s3 = q or r

                kb_value = s1 and s2 and s3

                print(
                    f"{q}\t{p}\t{r}\t{s1}\t{s2}\t{s3}\t{kb_value}"
                )

    print("-" * 75)


# Execute program
symbols = ['Q', 'P', 'R']

generate_truth_table()

result = tt_entails(KB, alpha, symbols)

print("\nENTAILMENT RESULT")
print("Does KB entail R?", result)


```
