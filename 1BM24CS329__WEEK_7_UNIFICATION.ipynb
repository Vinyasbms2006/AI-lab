```python

def is_variable(x):
    return isinstance(x, str) and x.islower()


def substitute(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, tuple):
        return tuple(substitute(arg, subst) for arg in term)

    return term


def occurs_check(var, term, subst):
    term = substitute(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, arg, subst)
                   for arg in term[1:])

    return False


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Main program

term1 = ("Eats", "x", "Riya")
term2 = ("Eats", "Apple", "y")

print("Term 1:", term1)
print("Term 2:", term2)

result = unify(term1, term2)

if result is not None:
    print("\nUnification Successful!")
    print("Most General Unifier (MGU):")

    for var, value in result.items():
        print(var, "=", value)

else:
    print("\nUnification Failed!")

```
