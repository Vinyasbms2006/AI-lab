```python
import random

def calculate_cost(state):
    """Cost = number of attacking pairs of queens"""
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def get_neighbors(state):
    """Generate all neighbors by moving one queen in its column"""
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if row != state[col]:
                neighbor = state.copy()
                neighbor[col] = row
                neighbors.append(neighbor)
    return neighbors

def hill_climbing_with_restart(n, max_restarts=100, plateau_patience=20):
    """Hill climbing with random restarts when stuck in local minima or plateau"""
    for restart in range(max_restarts):
        current = [random.randint(0, n-1) for _ in range(n)]
        plateau_counter = 0

        while True:
            cost = calculate_cost(current)
            print(f"Restart {restart}, Cost {cost}, State {current}")

            if cost == 0:
                print(" Solution found!")
                return current

            neighbors = get_neighbors(current)
            best = min(neighbors, key=calculate_cost)
            best_cost = calculate_cost(best)

            if best_cost >= cost:
                print("Local minimum or plateau reached. Restarting...")
                break

            if best_cost == cost:
                plateau_counter += 1
                if plateau_counter >= plateau_patience:
                    print("Plateau detected. Restarting...")
                    break

            current = best

    print("No solution found after max restarts.")
    return None


solution = hill_climbing_with_restart(4)

if solution:
    print("Final Solution:", solution)
    print("Final Cost:", calculate_cost(solution))


```
